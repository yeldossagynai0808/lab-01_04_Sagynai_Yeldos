```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np


days = ["дүйсенбі", "сейсенбі", "сәрсенбі", "бейсенбі", "жұма", "сенбі", "жексенбі"]
euro = [430, 505, 455, 510, 495, 415, 515]
usd = [480, 510, 525, 480, 515, 425, 400]


df_comparison = pd.DataFrame({
    'Күн': days,
    'EUR/KZT': euro,
    'USD/KZT': usd
})

print("Еуро және АҚШ долларының бағамдарын салыстыру кестесі:")
display(df_comparison)

plt.figure(figsize=(14, 8))

bar_width = 0.35
r = np.arange(len(df_comparison['Күн']))
```

```python
plt.figure(figsize=(14, 8))

bar_width = 0.35
r = np.arange(len(df_comparison['Күн']))


plt.bar(r - bar_width/2, df_comparison['EUR/KZT'],
        color='skyblue', width=bar_width, label='EUR/KZT')


plt.bar(r + bar_width/2, df_comparison['USD/KZT'],
        color='lightcoral', width=bar_width, label='USD/KZT')


plt.title('Еуро мен АҚШ долларының KZT-ға шаққандағы бағамдарын салыстыру', fontsize=16)
plt.xlabel('Күн', fontsize=12)
plt.ylabel('Бағам (KZT)', fontsize=12)

plt.xticks(r, df_comparison['Күн'], rotation=45, ha='right')


for i in range(len(df_comparison['Күн'])):
    plt.text(r[i] - bar_width/2,
             df_comparison['EUR/KZT'][i],
             f"{df_comparison['EUR/KZT'][i]:.0f}",
             ha='center', va='bottom', fontsize=9)

    plt.text(r[i] + bar_width/2,
             df_comparison['USD/KZT'][i],
             f"{df_comparison['USD/KZT'][i]:.0f}",
             ha='center', va='bottom', fontsize=9)

plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np


days = ["дүйсенбі", "сейсенбі", "сәрсенбі", "бейсенбі", "жұма", "сенбі", "жексенбі"]
euro = [430, 505, 455, 510, 495, 415, 515]


df_pie_chart = pd.DataFrame({
    'Күн': days,
    'EUR/KZT': euro
})

display(df_pie_chart) 
```

```python
plt.figure(figsize=(10, 8))

def func(pct, allvals):
    absolute = int(pct / 100. * np.sum(allvals))
    return f"{pct:.1f}%\n({absolute:d})"

wedges, texts, autotexts = plt.pie(
    df_pie_chart['EUR/KZT'],
    labels=df_pie_chart['Күн'],
    autopct=lambda pct: func(pct, df_pie_chart['EUR/KZT']),
    startangle=90,
    pctdistance=0.85,
    colors=plt.cm.Paired.colors
)

plt.title('Күндер бойынша EUR/KZT бағамының үлесі', fontsize=16)

plt.axis('equal')  


plt.setp(autotexts, size=10, weight="bold")
plt.setp(texts, size=11)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np


programs = ["Жаңалықтар", "Фильмдер", "Спорт", "Музыка", "Білім беру"]
hours = [3, 5, 2, 1, 2]   # қанша уақыт қарайды


df_tv = pd.DataFrame({
    'Бағдарлама': programs,
    'Сағат': hours
})

print(df_tv)

# График
plt.figure(figsize=(10, 8))

def func(pct, allvals):
    absolute = int(pct / 100. * np.sum(allvals))
    return f"{pct:.1f}%\n({absolute} сағ)"

plt.pie(
    df_tv['Сағат'],
    labels=df_tv['Бағдарлама'],
    autopct=lambda pct: func(pct, df_tv['Сағат']),
    startangle=90,
    pctdistance=0.85
)

plt.title("Теледидар бағдарламалары бойынша уақыт үлесі", fontsize=16)
plt.axis('equal')

plt.show()
```
